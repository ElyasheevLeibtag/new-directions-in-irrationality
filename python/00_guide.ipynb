## The three numerical conclusions

The subject notebooks establish or reproduce the computational ingredients of the rate estimates. This guide checks only the **final rational arithmetic**, using the paper's rounded bounds, not a separate proof of those estimates.

Write $|\alpha-p_n/q_n|\le e^{-gn+o(n)}$ and use an integer denominator representative of growth at most $e^{hn+o(n)}$. Under the paper's nondegeneracy and irrationality-measure hypotheses,
\[
\delta\ge\frac gh-1,\qquad \mu(\alpha)\le\frac{g}{g-h}.
\]
The lower bound on $g$ and upper bound on $h$ give the conservative consequences below.

```python
from final_bounds import verify_final_rounding
from IPython.display import Markdown, display
rows = verify_final_rounding()
lines = ["| Target | g lower | h upper | delta lower | mu upper | Paper claim |", "|---|---:|---:|---:|---:|---:|"]
for row in rows:
    lines.append(f"| {row['target']} | {row['g']} | {row['h']} | {row['delta_decimal']:.10f} | {row['mu_decimal']:.10f} | < {row['mu_claim']} |")
display(Markdown("\n".join(lines)))
print("All three final bounds pass exact rational comparison.")
```

For the joint result, the rounded statement $\delta>0.015$ alone would be insufficient for $\mu<66.3$. The calculation uses $g>112.8075$ and $h<111.105$, retaining the information needed for that conclusion.
